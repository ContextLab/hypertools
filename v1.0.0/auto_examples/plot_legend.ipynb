
# Generating a legend

An example of how to use the `legend` kwarg to generate a legend.


```python
# Code source: Andrew Heusser
# License: MIT

# import
import hypertools as hyp

# load example data
data = hyp.load('weights_sample')

# plot
hyp.plot(data, fmt='.', legend=['Group A', 'Group B', 'Group C'])
```
